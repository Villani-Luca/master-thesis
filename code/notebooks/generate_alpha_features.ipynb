# Data Generation for Alpha158 and Alpha360
Questo notebook mostra come chiamare le funzioni per generare le feature `Alpha158` e `Alpha360` sui dati di prezzo utilizzando gli script forniti nel framework FinBench.

```python
import os
import sys
import pandas as pd

# Aggiungiamo la directory Evaluation al path in modo da poter importare i moduli di FinBench
# Gestiamo il caso in cui il notebook venga eseguito dalla root del workspace o dalla cartella notebooks
if os.path.basename(os.getcwd()) == 'notebooks':
    finbench_eval_path = os.path.abspath('../finbench/Evaluation')
else:
    finbench_eval_path = os.path.abspath('./code/finbench/Evaluation')

if finbench_eval_path not in sys.path:
    sys.path.append(finbench_eval_path)

from features.alpha158 import calculate_alpha158
from features.alpha360 import calculate_alpha360
from utils.storage import read_stock_dataset
```

```python
universe = 'sp500' # Sostituire con l'universo/dataset desiderato (es. 'sp500', 'nasdaq100', 'dji', ecc.)
data_path = os.path.join(finbench_eval_path, 'data', universe, f'{universe}.csv')

if os.path.exists(data_path):
    print(f'Caricamento dataset {universe} da: {data_path}')
    prices = read_stock_dataset(data_path)
    display(prices.head())
else:
    print(f'Dataset non trovato in: {data_path}\nAssicurati di aver scaricato i dati di prezzo prima di procedere.')
    # Decommenta le due righe sotto se hai un mock dataframe per testare le funzioni 
    # import numpy as np
    # prices = pd.DataFrame(np.random.randn(100, 5), columns=['adj_open', 'adj_high', 'adj_low', 'adj_close', 'volume'])
```
