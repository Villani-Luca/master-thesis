# Data Generation for Alpha158 and Alpha360
Questo notebook carica i dataset già presenti nel repository e verifica i file `Alpha158` e `Alpha360` disponibili per l'universo scelto.

```python
import os
from pathlib import Path

import pandas as pd

# Risoluzione robusta del percorso del repo, indipendentemente da dove viene lanciato il notebook
cwd = Path.cwd().resolve()
project_root = None
data_root = None

for candidate in [cwd, cwd.parent, cwd.parent.parent]:
    if (candidate / "code" / "data").exists():
        project_root = candidate
        data_root = candidate / "code" / "data"
        break
    if candidate.name == "code" and (candidate / "data").exists():
        project_root = candidate.parent
        data_root = candidate / "data"
        break

if project_root is None:
    project_root = cwd
    data_root = cwd / "code" / "data"

print(f"Project root: {project_root}")
print(f"Data root: {data_root}")
```

```python
universe = "sp500"  # Sostituire con l'universo desiderato: sp500, nasdaq100, dji, sx5e, sxxp, ecc.
data_dir = data_root / universe
price_path = data_dir / f"{universe}.csv"
alpha158_path = data_dir / f"{universe}_alpha158.csv"
alpha360_path = data_dir / f"{universe}_alpha360.csv"

print(f"Dataset folder: {data_dir}")
for path in [price_path, alpha158_path, alpha360_path]:
    print(f"{path.name}: {'OK' if path.exists() else 'MISSING'}")

if price_path.exists():
    prices = pd.read_csv(price_path)
    display(prices.head())
    print(f"Prices shape: {prices.shape}")
else:
    print(f"Dataset dei prezzi non trovato: {price_path}")
```

```python
# Caricamento di Alpha158
if alpha158_path.exists():
    df_alpha158 = pd.read_csv(alpha158_path)
    print(f"Alpha158 caricato da: {alpha158_path}")
    display(df_alpha158.head())
    display(df_alpha158.shape)
else:
    print(f"File Alpha158 non trovato: {alpha158_path}")
    print("Se vuoi generarlo, assicurati che i dati prezzi siano disponibili e che il framework FinBench sia stato aggiunto al progetto.")
```

```python
# Caricamento di Alpha360
if alpha360_path.exists():
    df_alpha360 = pd.read_csv(alpha360_path)
    print(f"Alpha360 caricato da: {alpha360_path}")
    display(df_alpha360.head())
    display(df_alpha360.shape)
else:
    print(f"File Alpha360 non trovato: {alpha360_path}")
    print("Se vuoi generarlo, assicurati che i dati prezzi siano disponibili e che il framework FinBench sia stato aggiunto al progetto.")
```
